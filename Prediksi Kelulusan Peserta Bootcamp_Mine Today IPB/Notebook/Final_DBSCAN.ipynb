```python
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

df_combined = pd.concat([df_train, df_test], ignore_index=True)

fitur_kinerja = [
    'jumlah_kehadiran',
    'jumlah_mini_project_dikerjakan',
    'jumlah_quiz_dikerjakan',
    'rata_rata_score',
    'Score_x',
    'Score_y',
    'Kualitas materi',
    'Kualitas trainer',
    'Status_Fresh Graduates',
    'Status_Mahasiswa',
    'Status_Pekerja aktif',
    'Status_Umum',
    'Siapa trainer pada hari ini ?_ln',
    'Siapa trainer pada hari ini ?_mdn',
    'Siapa trainer pada hari ini ?_mf',
    'Siapa trainer pada hari ini ?_ra',
    'Pilihan Jadwal Kelas_Bootcamp Data Science For Beginner Batch 9 - Mulai April 2024 (Kelas Terdekat)',
    'Bootcamp_Online',
    'Score'
]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_combined[fitur_kinerja])

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

dbscan = DBSCAN(eps=0.5, min_samples=5)
cluster_dbscan = dbscan.fit_predict(X_pca)

df_combined['Cluster_DBSCAN'] = cluster_dbscan

df_test_result_dbscan = df_combined.iloc[len(df_train):].copy()
n_clusters = len(set(cluster_dbscan)) - (1 if -1 in cluster_dbscan else 0)

if n_clusters == 2:
    df_test_result_dbscan['Prediksi_Lulus'] = df_test_result_dbscan['Cluster_DBSCAN'].apply(lambda x: 1 if x == 0 else 0)
else:
    df_test_result_dbscan['Prediksi_Lulus'] = 0

hasil_prediksi = df_test_result_dbscan[['Prediksi_Lulus']].reset_index(drop=True)
```

```python
mask = cluster_dbscan != -1
X_valid = X_scaled[mask]
labels_valid = cluster_dbscan[mask]

score = silhouette_score(X_valid, labels_valid)
print(f"Silhouette Score (tanpa noise): {score:.4f}")
```
